```python
import sympy as sp
import numpy as np
from sympy import exp,Heaviside,plot,fourier_series
from sympy.abc import t,f
import matplotlib.pyplot as plt
import scienceplots 
plt.style.use(["science","no-latex"])
```

```python
g=exp(-t/2)*(Heaviside(t)-Heaviside(t-sp.pi))
```

```python
plot(g,g.subs(t,t-sp.pi),g.subs(t,t+sp.pi))
plt.show()
```

```python
sol=fourier_series(g,(t,0,sp.pi))
sol.truncate(10)
```
