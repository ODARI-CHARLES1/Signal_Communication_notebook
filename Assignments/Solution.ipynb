```python
import numpy as np
import sympy as sp
from sympy import Heaviside,exp,fourier_transform,plot,Abs,lambdify
from sympy.abc import t,f
import matplotlib.pyplot as plt
import scienceplots

plt.style.use(["science","no-latex"])
```

```python
g=1e-3*exp(-1000*t)*Heaviside(t)
G=fourier_transform(g,t,f)
G_num=lambdify(f,Abs(G),modules=["numpy"])
```

```python
f = np.linspace(-500, 500, 1000)

g_num = G_num(f)

cut_off = G_num(0) / np.sqrt(2)

plt.plot(f, g_num, label='|G(f)|')

# Vertical line at cutoff frequency
plt.axvline(159.15, color='r', linestyle='--', label='fc')
plt.axvline(-159.15, color='r', linestyle='--', label='fc')

# Horizontal line at -3 dB level
plt.axhline(cut_off, color='g', linestyle='--', label='-3 dB')

plt.xlabel("Frequency (Hz)")
plt.ylabel("|G(f)|")
plt.grid(True)
plt.legend()

plt.show()
```
