```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
fs = 1000                    # Sampling frequency (Hz)
T = 2                        # Duration (seconds)
t = np.linspace(0, T, fs*T)

f = 5                      
signal = np.sin(2*np.pi*f*t)

noise = 0.4*np.random.randn(len(t))
x = signal + noise
autocorr = np.correlate(x, x, mode='full')

autocorr = autocorr / np.max(autocorr)

# Lags
lags = np.arange(-len(x)+1, len(x))
plt.figure(figsize=(12,8))

plt.subplot(2,1,1)
plt.plot(t, x)
plt.title("Noisy Sine Wave")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.grid(True)

plt.subplot(2,1,2)
plt.plot(lags, autocorr)
plt.title("Autocorrelation")
plt.xlabel("Lag (samples)")
plt.ylabel("Normalized Correlation")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python

noise = np.random.randn(1000)
autocorr = np.correlate(noise, noise, mode='full')
autocorr /= np.max(autocorr)

lags = np.arange(-len(noise)+1, len(noise))

plt.figure(figsize=(10,4))
plt.plot(lags, autocorr)
plt.title("Autocorrelation of Random Noise")
plt.xlabel("Lag")
plt.ylabel("Normalized Correlation")
plt.grid(True)
plt.show()
```

```python
t = np.linspace(-10, 10, 1000)
x = np.zeros_like(t)
y=np.zeros_like(x)
x[(t >= -2) & (t <= 2)] = 1
y[(t>=-1)&(t<=1)]=1
Rxx = np.correlate(x, y, mode='full')

Rxx = Rxx / np.max(Rxx)

# Lag axis
dt = t[1] - t[0]
lags = np.arange(-len(x)+1, len(x)) * dt

plt.figure(figsize=(12,6))

plt.subplot(3,1,1)
plt.plot(t, x, linewidth=2)
plt.title("Rectangular Pulse")
plt.xlim([-4,4])
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.grid(True)

plt.subplot(3,1,3)
plt.plot(lags, Rxx, linewidth=2)
plt.xlim([-4,4])
plt.title("Autocorrelation of the Rectangular Pulse")
plt.xlabel("Lag")
plt.ylabel("Normalized Correlation")
plt.grid(True)

plt.subplot(3,1,2)
plt.plot(t,y,linewidth=2)
plt.xlim([-4,4])
plt.title("Rectangular pulse with $\tau$=2")
plt.ylabel("Amplitude")
plt.xlabel("Time")
plt.grid()


plt.tight_layout()
plt.show()
```

```python
# Time vector
t = np.linspace(0, 2*np.pi, 1000)

# Orthogonal signals
x = np.sin(t)
y = np.cos(t)

# Cross-correlation
corr = np.correlate(x, y, mode='full')

# Normalize
corr = corr / np.max(np.abs(corr))

lags = np.arange(-len(x)+1, len(x))

plt.figure(figsize=(12,6))

plt.subplot(2,1,1)
plt.plot(t, x, label='sin(t)')
plt.plot(t, y, label='cos(t)')
plt.title("Orthogonal Signals")
plt.legend()
plt.grid(True)

plt.subplot(2,1,2)
plt.plot(lags, corr)
plt.title("Cross-Correlation between sin(t) and cos(t)")
plt.xlabel("Lag")
plt.ylabel("Correlation")
plt.grid(True)

plt.tight_layout()
plt.show()

print("Correlation at zero lag =", np.dot(x, y))
```
