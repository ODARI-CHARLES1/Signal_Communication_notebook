```python
import sympy as sp
import matplotlib.pyplot as plt
import scienceplots 
import numpy as np
from ipywidgets import interact, interactive, fixed, interact_manual
import ipywidgets as widgets
```

```python
def f(x):
    t=np.linspace(0,0.8,1000)
    y=5*(1+x*3*np.sin(100*t))*np.cos(400*t)
    plt.plot(t,y)
    plt.axhline(y=0)
    return x
```

```python
interact(f,x=(0.0, 1.0, 0.1));
```
