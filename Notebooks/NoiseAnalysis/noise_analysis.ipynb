```python
import numpy as np
from numpy import sin,pi,exp
import pandas as pd
import matplotlib.pyplot as plt
```

```python
t=np.linspace(0,1,1000);
y=sin(2*pi*10*t)+exp(-4*t)+5e-2;

# Structure the data into a dictionary
noise_data = {"Time": t, "Noise_data": y}

# Create a DataFrame and save it to a CSV file
df = pd.DataFrame(noise_data)
df.to_csv("Noise_data.csv", index=False)

#Determine the average and ac_component
noise_mean=np.mean(y)
ac_component=y-noise_mean

#plotting the signal 
plt.plot(t,y,'b',label="Noise Signal")
plt.axhline(y=noise_mean,label="Noise Mean")

plt.grid(True,linestyle='--',alpha=0.5,axis='y')
plt.legend()
plt.title("Signal analysis plots")
plt.xlabel("t")
plt.ylabel("Amplitude")
plt.show()
```

```python
plt.plot(t,ac_component,'g',label="Noise Ac component")
plt.title("Noise Ac Component")
plt.xlabel("t")
plt.grid(True,linestyle='--',alpha=0.5,axis='y')
plt.ylabel("Amplitude")
plt.show()
```
