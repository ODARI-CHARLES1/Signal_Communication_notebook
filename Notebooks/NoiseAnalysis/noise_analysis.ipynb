```python
import numpy as np
from numpy import sin,pi,exp
import pandas as pd
import matplotlib.pyplot as plt
```

```python
t=np.linspace(0,1,1000);
y=sin(2*pi*10*t)+exp(-4*t)+5e2;


noise_mean=np.mean(y)
ac_component=y-noise_mean

#plotting the signal 
plt.plot(t,y,'b',label="Noise Signal")
plt.axhline(y=noise_mean,label="Noise Mean")

plt.grid(True,linestyle='--',alpha=0.5,axis='y')
plt.legend()
plt.title("Signal analysis plots")
plt.xlabel("t")
plt.ylabel("Amplitude")
plt.show()
```

```python
plt.plot(t,ac_component,'g',label="Noise Ac component")
```
