```python
import sympy as sp
from sympy.abc import t,f
import scienceplots
import matplotlib.pyplot as plt
from sympy import Heaviside,plot,exp,fourier_transform

plt.style.use(["science","no-latex"])
```

```python
y=sp.exp(-t)*Heaviside(t)
Y=fourier_transform(y,t,f)
Y_mag=sp.Abs(Y)
Y_phase=sp.arg(Y)
```

```python
plot(y,xlim=[0,10])
plt.show()
```

```python
plot(Y_mag,xlim=[-2,2],title="Magnitude plot for fourier transform of $e^{-t}$")
plt.show()
```

```python
plot(Y_phase,xlim=[-5,5],title="Phase plot for fourier tranform of $e^{-t}$")
plt.show()
```

```python
g=sp.exp(-sp.Abs(2e-2*t))
plot(g)
```

```python
G=fourier_transform(g,t,f)
plot(G,xlim=[-2,2])
plt.show()
```
