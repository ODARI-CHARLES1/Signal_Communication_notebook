### Problem 1: Multiple Signal Transformation

Given:
$$
x(t)=
\begin{cases}
2-t, & 1 \le t \le 2 \\
1, & 2 < t \le 4 \\
0, & \text{otherwise}
\end{cases}
$$



```python
#define the function
y = Piecewise(
    (2 - t, (t >= 1) & (t <= 2)),
    (1, (t > 2) & (t <= 4)),
    (0, True)
)
```

```python
def plot_func(y_func,lim=[]):
    sp.plot(y_func,xlim=lim)
    plt.show()
```

```python
plot_func(y,lim=[0,6]) #original function
```

```python
plot_func(y.subs(t,-t),lim=[-6,0]) #substitute t , -t
```

```python
plot_func(y.subs(t,t-2),lim=[0,8]) #Substitute t,t-2
          
```

```python
plot_func(y.subs(t,2*t),lim=[0,4])
```
