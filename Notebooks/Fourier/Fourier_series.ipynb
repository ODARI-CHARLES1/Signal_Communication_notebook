```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
import scienceplots
import warnings
from sympy import Heaviside
warnings.filterwarnings("ignore")
plt.style.use(["science","no-latex"])
```

```python
from sympy.abc import f,t
```

```python
g=Heaviside(t)-Heaviside(t-2)
sp.plot(g,g.subs(t,(t-4)),g.subs(t,(t-8)),xlim=[-1,15],title="Rectangular plot")
plt.show()
```

```python
#Get fourier series of a signal,being odd function
sol=sp.fourier_series(g,(t,-2,2))
sol.truncate(10)
```

```python
#Get Fourier series of a signal, being even function
sol=sp.fourier_series(g.subs(t,t+1),(t,-2,2))
sol.truncate(10)
```
